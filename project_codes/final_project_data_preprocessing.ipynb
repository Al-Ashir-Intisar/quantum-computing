```python
# prompt: read in a .txt file and count how many characters are in that file?

def count_characters(filepath):
  """Counts the number of characters in a text file.

  Args:
    filepath: The path to the text file.

  Returns:
    The number of characters in the file, or -1 if the file does not exist.
  """
  try:
    with open(filepath, 'r') as file:
      contents = file.read()
      return len(contents)
  except FileNotFoundError:
    print(f"Error: File not found at {filepath}")
    return -1

# Example usage (replace with the actual file path)
file_path = os.path.join(negative_dir, '0.txt')
char_count = count_characters(file_path)

if char_count != -1:
  print(f"The file contains {char_count} characters.")
```

```python

def generate_random_numbers(count, min_val, max_val):
  """Generates a list of random numbers.

  Args:
    count: The number of random numbers to generate.
    min_val: The minimum value for the random numbers.
    max_val: The maximum value for the random numbers.

  Returns:
    A list of random numbers.
  """
  random_numbers = [random.randint(min_val, max_val) for _ in range(count)]
  return random_numbers

```

```python
def random_sample(filepath, sample_size=200):
  negative_dir = os.path.join(filepath, 'sequence/negative')
  positive_dir = os.path.join(filepath, 'sequence/positive')

  # Count the number of files in the extracted directory
  num_files_negative = len([f for f in os.listdir(negative_dir) if os.path.isfile(os.path.join(negative_dir, f))])
  num_files_positive = len([f for f in os.listdir(positive_dir) if os.path.isfile(os.path.join(positive_dir, f))])
  print(f"Number of files in {negative_dir}: {num_files_negative}")
  print(f"Number of files in {positive_dir}: {num_files_positive}")
  random_idx_negative = generate_random_numbers(sample_size, 0, num_files_negative - 1)
  random_idx_positive = generate_random_numbers(sample_size, 0, num_files_positive - 1)

  data = []

  for i in random_idx_negative:
      file_path = os.path.join(negative_dir, f'{i}.txt')
      char_count = count_characters(file_path)
      if char_count != -1:
          try:
            with open(file_path, 'r') as file:
                content = file.read()
                data.append([i, char_count,content, 'negative'])
          except Exception as e:
            print(f"Error reading file {file_path}: {e}")

  for i in random_idx_positive:
      file_path = os.path.join(positive_dir, f'{i}.txt')
      char_count = count_characters(file_path)
      if char_count != -1:
          try:
            with open(file_path, 'r') as file:
                content = file.read()
                data.append([i, char_count, content, 'positive'])
          except Exception as e:
            print(f"Error reading file {file_path}: {e}")

  df = pd.DataFrame(data, columns=['Index', 'Length', 'Text', 'Label'])
  print(df.head())
  return df
```

```python
#random seed for reproducibility and sampling 200 samples from each class
random.seed(5402)
sample_df = random_sample(extract_dir, 400)
```

```python
# using one hot encoding to convert dna sequence to numeric representation

def one_hot_encode(sequence):
    encoding_map = {'A': [1, 0, 0, 0], 'C': [0, 1, 0, 0], 'G': [0, 0, 1, 0], 'T': [0, 0, 0, 1]}
    encoded_sequence = [encoding_map.get(char, [0, 0, 0, 0]) for char in sequence]  # Default to [0,0,0,0] for unknown chars
    return np.ravel(encoded_sequence)

# Example usage (assuming 'Text' column contains DNA sequences)
sample_df['Encoded_Text'] = sample_df['Text'].apply(one_hot_encode)
print(sample_df.head())
```

```python
dna_encoded_matrix = np.zeros((len(sample_df), sample_df['Length'].max()*4))
dna_encoded_matrix.shape
```

```python
for i in range(len(sample_df['Length'])):
  if len(sample_df['Encoded_Text'][i]) < dna_encoded_matrix.shape[1]:
    dna_encoded_matrix[i, :len(sample_df['Encoded_Text'][i])] = sample_df['Encoded_Text'][i]
```

```python
matrix_df = pd.DataFrame(dna_encoded_matrix)
# matrix_df.to_csv('dna_encoded_matrix.csv', index=False)
```

```python
# Conducting Isomap for dimensionality reduction
# Use your high-dimensional input
X = matrix_df.values.astype(float)

# Step 1: Compute full pairwise distances in the original space
D_original = pairwise_distances(X)

# Step 2: Run Isomap to get reduced dimensions (e.g., 10 for explanation analysis)
n_components = 64
isomap = Isomap(n_neighbors=5, n_components=n_components)
X_iso = isomap.fit_transform(X)

# Step 3: Compute pairwise distances in reduced space
D_reduced = pairwise_distances(X_iso)

# Step 4: Compute how much variance is preserved
# Flatten the upper triangle (excluding diagonal)
orig_flat = D_original[np.triu_indices_from(D_original, k=1)]
reduced_flat = D_reduced[np.triu_indices_from(D_reduced, k=1)]

# Compute R^2 as an approximation of variance preserved
correlation_matrix = np.corrcoef(orig_flat, reduced_flat)
r_squared = correlation_matrix[0, 1] ** 2

print(f"Approximate variance explained by {n_components} Isomap components: {r_squared * 100:.2f}%")
```

```python
X_reduced_isomap_df = pd.DataFrame(X_iso, columns=[f"iso_{i}" for i in range(n_components)])
X_reduced_isomap_df.head(2)
```

```python
# conducting Principal component analysis to see if it explains more variation in the dataset for lower features.
# Use your high-dimensional input
X = matrix_df.values.astype(float)

# Step 1: Fit PCA
n_components = 64
pca = PCA(n_components=n_components)
X_pca = pca.fit_transform(X)

# Step 2: Explained variance per component
explained_variance_ratios = pca.explained_variance_ratio_

# Step 3: Total variance explained by first 64 components
total_explained_variance = np.sum(explained_variance_ratios)

print(f"Total variance explained by first {n_components} PCA components: {total_explained_variance * 100:.2f}%")
```

```python
X_reduced_pca_df = pd.DataFrame(X_pca, columns=[f"pca_{i}" for i in range(n_components)])
X_reduced_pca_df.head(2)
```
