```python
# prompt: read in a .txt file and count how many characters are in that file?

def count_characters(filepath):
  """Counts the number of characters in a text file.

  Args:
    filepath: The path to the text file.

  Returns:
    The number of characters in the file, or -1 if the file does not exist.
  """
  try:
    with open(filepath, 'r') as file:
      contents = file.read()
      return len(contents)
  except FileNotFoundError:
    print(f"Error: File not found at {filepath}")
    return -1

# Example usage (replace with the actual file path)
file_path = os.path.join(negative_dir, '0.txt')
char_count = count_characters(file_path)

if char_count != -1:
  print(f"The file contains {char_count} characters.")
```

```python

def generate_random_numbers(count, min_val, max_val):
  """Generates a list of random numbers.

  Args:
    count: The number of random numbers to generate.
    min_val: The minimum value for the random numbers.
    max_val: The maximum value for the random numbers.

  Returns:
    A list of random numbers.
  """
  random_numbers = [random.randint(min_val, max_val) for _ in range(count)]
  return random_numbers

```

```python
def random_sample(filepath, sample_size=200):
  negative_dir = os.path.join(filepath, 'sequence/negative')
  positive_dir = os.path.join(filepath, 'sequence/positive')

  # Count the number of files in the extracted directory
  num_files_negative = len([f for f in os.listdir(negative_dir) if os.path.isfile(os.path.join(negative_dir, f))])
  num_files_positive = len([f for f in os.listdir(positive_dir) if os.path.isfile(os.path.join(positive_dir, f))])
  print(f"Number of files in {negative_dir}: {num_files_negative}")
  print(f"Number of files in {positive_dir}: {num_files_positive}")
  random_idx_negative = generate_random_numbers(sample_size, 0, num_files_negative - 1)
  random_idx_positive = generate_random_numbers(sample_size, 0, num_files_positive - 1)

  data = []

  for i in random_idx_negative:
      file_path = os.path.join(negative_dir, f'{i}.txt')
      char_count = count_characters(file_path)
      if char_count != -1:
          try:
            with open(file_path, 'r') as file:
                content = file.read()
                data.append([i, char_count,content, 'negative'])
          except Exception as e:
            print(f"Error reading file {file_path}: {e}")

  for i in random_idx_positive:
      file_path = os.path.join(positive_dir, f'{i}.txt')
      char_count = count_characters(file_path)
      if char_count != -1:
          try:
            with open(file_path, 'r') as file:
                content = file.read()
                data.append([i, char_count, content, 'positive'])
          except Exception as e:
            print(f"Error reading file {file_path}: {e}")

  df = pd.DataFrame(data, columns=['Index', 'Length', 'Text', 'Label'])
  print(df.head())
  return df
```
